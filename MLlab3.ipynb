```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1, 10, 10)
y1 = x
y2 = x**2

# 1. Change legend position
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.legend(loc='upper left')
plt.show()

# 2. Change legend font size
plt.plot(x, y1, label='Line 1')
plt.legend(fontsize=14)
plt.show()

# 3. Change vertical spacing between legend entries
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.legend(labelspacing=1.5)
plt.show()

# 4. Use several columns in the legend
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.legend(ncol=2)
plt.show()

# 5. One legend for all subplots
fig, axes = plt.subplots(1, 2)
axes[0].plot(x, y1, label='Linear')
axes[1].plot(x, y2, label='Quadratic')
handles, labels = axes[0].get_legend_handles_labels()
h2, l2 = axes[1].get_legend_handles_labels()
fig.legend(handles + h2, labels + l2, loc='upper center')
plt.tight_layout()
plt.show()

# 6. Legend with a coloured frame
plt.plot(x, y1, label='Line 1')
plt.legend(frameon=True, facecolor='lightyellow', edgecolor='red')
plt.show()

# 7. Place legend outside the plot
plt.plot(x, y1, label='Line 1')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()

# 8. Remove legend border
plt.plot(x, y1, label='Line 1')
plt.legend(frameon=False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

# 1. Draw a horizontal line
plt.plot(x, y)
plt.axhline(y=0.5, color='red')
plt.show()

# 2. Several horizontal lines with different styles
plt.plot(x, y)
plt.axhline(0.5, color='red', linestyle='--', linewidth=2)
plt.axhline(0, color='blue', linestyle=':', linewidth=3)
plt.axhline(-0.5, color='green', linestyle='-.', linewidth=1)
plt.show()

# 3. Draw a vertical line
plt.plot(x, y)
plt.axvline(x=5, color='red')
plt.show()

# 4. Several vertical lines with different styles
plt.plot(x, y)
plt.axvline(2, color='red', linestyle='--', linewidth=2)
plt.axvline(5, color='blue', linestyle=':', linewidth=3)
plt.axvline(8, color='green', linestyle='-.', linewidth=1)
plt.show()

# 5. Change line opacity
plt.plot(x, y, alpha=0.3)
plt.show()

# 6. Increase line thickness
plt.plot(x, y, linewidth=4)
plt.show()

# 7. Fill between two lines
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.fill_between(x, np.sin(x), np.cos(x), alpha=0.3)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ['A', 'B', 'C', 'D']
values = [25, 40, 15, 30]

# 1. Horizontal bar chart
plt.barh(categories, values)
plt.xlabel('Value')
plt.show()

# 2. Show the value of each bar
bars = plt.bar(categories, values)
plt.bar_label(bars)
plt.show()

# 3. Annotate each bar manually
bars = plt.bar(categories, values)
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(bar.get_height()),
        ha='center',
        va='bottom'
    )
plt.show()

# 4. Annotate a grouped bar chart
x = np.arange(len(categories))
width = 0.35
fig, ax = plt.subplots()
bars1 = ax.bar(x - width/2, values, width, label='Group 1')
bars2 = ax.bar(x + width/2, [20, 30, 25, 35], width, label='Group 2')
ax.bar_label(bars1)
ax.bar_label(bars2)
ax.set_xticks(x, categories)
ax.legend()
plt.show()

# 5. Housing class: count categories from your loaded Ames dataset
# MSSubClass is the closest matching column in train.csv.
counts = Housing['MSSubClass'].value_counts().sort_values(ascending=False)
counts.plot(kind='bar')
plt.title('Housing Class Counts')
plt.xlabel('MSSubClass')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Print the category order from largest to smallest
print('Sequence:', ''.join(str(int(c)) for c in counts.index))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 11)
y = x + np.random.rand(10)

# 1. Add a legend
plt.scatter(x, y, label='Observations')
plt.legend()
plt.show()

# 2. Connect scatter points with a line
plt.plot(x, y, marker='o', label='Observations')
plt.legend()
plt.show()

# 3. Use several colours
colours = ['red', 'blue', 'green', 'orange', 'purple',
           'red', 'blue', 'green', 'orange', 'purple']
plt.scatter(x, y, c=colours)
plt.show()

# 4. Increase point size
plt.scatter(x, y, s=150)
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['A', 'B', 'C', 'D']
values = [35, 25, 20, 20]

# 1. Draw a pie chart
plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.show()

# 2. Add borders around the wedges
plt.pie(
    values,
    labels=labels,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2}
)
plt.show()

# 3. Offset one wedge
explode = [0, 0.15, 0, 0]
plt.pie(values, labels=labels, explode=explode, autopct='%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

x = np.arange(1, 11)
y = x**2

# 1. Change transparency
plt.plot(x, y, alpha=0.4)
plt.show()

# 2. Change line colour
plt.plot(x, y, color='red')
plt.show()

# 3. Change title font size
plt.plot(x, y)
plt.title('My Plot', fontsize=18)
plt.show()

# 4. Change tick-label font size
plt.plot(x, y)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.show()

# 5. Change plot background colour
fig, ax = plt.subplots()
ax.set_facecolor('lightyellow')
ax.plot(x, y)
plt.show()

# 6. Generate a random colour
random_colour = '#' + ''.join(random.choices('0123456789ABCDEF', k=6))
plt.plot(x, y, color=random_colour)
plt.show()

# 7. Add text inside a plot
plt.plot(x, y)
plt.text(5, 50, 'Important point')
plt.show()

# 8. Add text using axes coordinates
plt.plot(x, y)
plt.text(0.5, 0.9, 'Plot annotation', transform=plt.gca().transAxes)
plt.show()

# 9. Change colour-bar size
data = np.random.rand(10, 10)
fig, ax = plt.subplots()
image = ax.imshow(data)
fig.colorbar(image, ax=ax, shrink=0.6)
plt.show()

# 10. Change figure size
plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.show()

# 11. Change axis-label font size
plt.plot(x, y)
plt.xlabel('X values', fontsize=16)
plt.ylabel('Y values', fontsize=16)
plt.show()

# 12. Hide axis tick labels
plt.plot(x, y)
plt.xticks([])
plt.yticks([])
plt.show()

# 13. Adjust axis-label positions
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('X values', labelpad=20)
ax.set_ylabel('Y values', labelpad=20)
plt.show()

# 14. Hide axes and remove whitespace
fig, ax = plt.subplots()
ax.plot(x, y)
ax.axis('off')
plt.margins(0)
plt.show()
```
