```python
import numpy as np

a = np.random.sample((10, 10))
b = np.random.sample((10, 10))

def mult(a, b):
    n, m, p = len(a), len(b), len(b[0])
    result = [[0] * p for _ in range(n)]
    for i in range(n):          # each row of a
        for j in range(p):      # each column of b
            for k in range(m):  # multiply and add up
                result[i][j] += a[i][k] * b[k][j]
    return result

slow = mult(a, b)
print(slow[0])   # just the first row
```

```python
import time

def np_mult(a, b):
    return a @ b      # NumPy's built-in grid multiplication

t = time.time()
slow = mult(a, b)
print("loops:", time.time() - t, "seconds")

t = time.time()
fast = np_mult(a, b)
print("numpy:", time.time() - t, "seconds")

print("same answer?", np.allclose(np.array(slow), fast))
```

```python
import numpy as np
import time

A = np.random.rand(10**6)

def sec_av(A):
    return np.cumsum(A) / np.arange(1, len(A) + 1)

def stupid_sec_av(A):
    S = [0 for i in range(len(A))]
    for i in range(len(A)):
        S[i] = A[i] + (S[i-1] if i > 0 else 0)   # running total
    for i in range(len(A)):
        S[i] = S[i] / (i + 1)                    # divide by how many so far
    return S

t = time.time(); S1 = sec_av(A);        print("numpy:", time.time() - t)
t = time.time(); S2 = stupid_sec_av(A); print("loops:", time.time() - t)

print("difference:", np.abs(S1 - np.array(S2)).sum())
```

```python
import numpy as np

def transformation(X, a=1):
    Y = X.copy()                  # X stays unchanged
    Y[1::2] = a                   # odd positions become a
    Y[::2] = X[::2][::-1] ** 3    # even positions: cubed, in reverse order
    return np.concatenate((X, Y))[::-1]   # join X and Y, then flip everything

def stupid_transformation(X, a=1):
    n = len(X)
    Y = [0] * n
    evens = [X[i] ** 3 for i in range(0, n, 2)][::-1]
    e = 0
    for i in range(n):
        if i % 2 == 1:
            Y[i] = a
        else:
            Y[i] = evens[e]
            e += 1
    return np.array(list(X) + Y)[::-1]

# small test
print(transformation(np.array([1., 2, 3, 4, 5])))

# both versions should agree
X = np.random.rand(1001)
print("difference:", np.abs(transformation(X) - stupid_transformation(X)).sum())
```

```python
import numpy as np

def cool_matrix_product(A, B):
    # None adds an extra direction, so every pair gets multiplied at once.
    # Then .sum(axis=1) adds up the pairs that belong to the same answer cell.
    return (A[:, :, None] * B[None, :, :]).sum(axis=1)

A = np.random.rand(4, 5)
B = np.random.rand(5, 3)

print(cool_matrix_product(A, B))
print("matches A @ B?", np.allclose(cool_matrix_product(A, B), A @ B))
```
