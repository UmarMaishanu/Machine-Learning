```python
nums = [1,2,3,4,5]
letters = ['a', 'b', 'c',]

x = nums.copy()
x[0:3] = letters
print("same size:        ", x)

x = nums.copy()
x[0:3] = ['X']
print("smaller:     ", x)

x = nums.copy()
x[0:1] = letters
print("bigger:      ", x)

x = nums.copy()
x[1:3] = []
print("empty (delete): ", x)

x = nums.copy()
x[1:4] = x[0:3]
print("overlap 1:     ", x)

x = nums.copy()
x[0:3] = x[2:5]
print("overlap 2:      ", x)




```

```python
# Task 1, way 1: loop + nested if
result_loop = []
for n in range(1, 101):
    if n ** 2 % 7 == 2:
        result_loop.append(n)
print(result_loop)

# Task 1, way 2: list comprehension
result_comp = [n for n in range(1, 101) if n ** 2 % 7 == 2]
print(result_comp)
print("same?", result_loop == result_comp)

# Bonus: n^2 mod 7 = 3
bonus = [n for n in range(1, 101) if n ** 2 % 7 == 3]
print("bonus:", bonus)

# Task 2: print 100 numbers separated by commas, no plain loops
print(', '.join([str(n) for n in range(1, 101)]))
```

```python
# Task 1: keep only "clean" words (letters only)
texts = ["hello w0rld this is c@t",
         "python3 is fun and e4sy",
         "ma#chine learning is cool"]

cleaned = [' '.join([w for w in t.split() if w.isalpha()]) for t in texts]
print(cleaned)

# Task 2
strings = ['agfkd.,f', 'Qksdf;sb&..', 'asdoo*', 'bgf...d', 're54()kj[]].']

# 2.1 how many periods in each string
counts = [s.count('.') for s in strings]
print(counts)

# 2.2 only strings with more than 2 periods
many_dots = [s for s in strings if s.count('.') > 2]
print(many_dots)
```
